# BTMP-JPEG 

```python
from PIL import Image
import os

input_directory = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/Adenoma/Adenoma'  # Replace with the path to your input directory
output_directory = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/JPEG/Adenoma/Adenoma'  # Replace with the path to the output directory

os.makedirs(output_directory, exist_ok=True)

for filename in os.listdir(input_directory):
    if filename.endswith('.bmp'):
        bmp_path = os.path.join(input_directory, filename)
        img = Image.open(bmp_path)
        
        # Resize the image to 150x150 pixels
        img_resized = img.resize((150, 150), Image.ANTIALIAS)
        
        # Save the resized image as a JPEG
        output_filename = os.path.splitext(filename)[0] + '.jpg'
        jpg_path = os.path.join(output_directory, output_filename)
        img_resized.save(jpg_path, 'JPEG')
        
        print(f"Converted and resized: {filename} -> {output_filename}")

print("Conversion and resizing complete.")
```

# Filter

Adenoma

```python
import cv2
import os
import pywt

# Define the directory where the original images are stored
img_dir = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/JPEG/Adenoma/Adenoma'

# Define the directory to save the filtered images
filtered_dir = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/JPEG/Filter/Adenoma'

# Create the directory if it doesn't exist
if not os.path.exists(filtered_dir):
    os.makedirs(filtered_dir)

# Loop through the images in the directory
for file_name in os.listdir(img_dir):
    
    # Load the image
    img_path = os.path.join(img_dir, file_name)
    img = cv2.imread(img_path)
    
    # Apply the Gaussian filter
    img_gaussian = cv2.GaussianBlur(img, (5, 5), 0)
    
    # Apply the median filter
    img_median = cv2.medianBlur(img, 5)
    
    
    # Save the filtered images in the directory
    cv2.imwrite(os.path.join(filtered_dir, 'gaussian_'+file_name), img_gaussian)
    cv2.imwrite(os.path.join(filtered_dir, 'median_'+file_name), img_median)
```

```python
import cv2
import os

# Set the input and output directory paths
input_dir = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/JPEG/Adenoma/Adenoma'
output_dir = 'D:/College Study Stuff/Project/Brain Tumor Web/Datasets/19915306/JPEG/Negative/Adenoma'

# Create the output directory if it doesn't exist
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Loop through each file in the input directory
for filename in os.listdir(input_dir):
    # Check if the file is an image
    if filename.endswith('.jpg') or filename.endswith('.jpeg') or filename.endswith('.png'):
        # Read the image
        img = cv2.imread(os.path.join(input_dir, filename))
        
        # Apply the negative filter
        neg_img = 255 - img
        
        # Save the output image to the output directory
        cv2.imwrite(os.path.join(output_dir, filename), neg_img)
```

Applying both

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread("D:/College Study Stuff/Project/Brain Tumor Web/Datasets/Braintumosparseclassification/Training/glioma_tumorgg (652).jpg")

# Apply a median filter
median_filtered = cv2.medianBlur(image, 5)  # 5 is the kernel size, adjust as needed

# Apply a Gaussian filter
gaussian_filtered = cv2.GaussianBlur(image, (5, 5), 0)  # (5, 5) is the kernel size, adjust as needed

# Apply both filters in sequence
median_gaussian_filtered = cv2.GaussianBlur(median_filtered, (5, 5), 0)

# Display the original image and the filtered images
plt.figure(figsize=(10, 4))

plt.subplot(1, 4, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))

plt.subplot(1, 4, 2)
plt.title('Median Filter')
plt.imshow(cv2.cvtColor(median_filtered, cv2.COLOR_BGR2RGB))

plt.subplot(1, 4, 3)
plt.title('Gaussian Filter')
plt.imshow(cv2.cvtColor(gaussian_filtered, cv2.COLOR_BGR2RGB))

plt.subplot(1, 4, 4)
plt.title('Median + Gaussian Filter')
plt.imshow(cv2.cvtColor(median_gaussian_filtered, cv2.COLOR_BGR2RGB))

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread("D:/College Study Stuff/Project/Brain Tumor Web/Datasets/Braintumosparseclassification/Training/glioma_tumor/gg (652).jpg", cv2.IMREAD_GRAYSCALE)

# Apply a Sobel filter
sobel_filtered_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=5)
sobel_filtered_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=5)
sobel_filtered = np.sqrt(sobel_filtered_x**2 + sobel_filtered_y**2)

# Apply a median filter
median_filtered = cv2.medianBlur(image, 5)  # Adjust the kernel size as needed

# Apply both filters in sequence
sobel_median_filtered = np.sqrt(sobel_filtered**2 + median_filtered**2)

# Display the original image and the filtered images
plt.figure(figsize=(12, 4))

plt.subplot(1, 4, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')

plt.subplot(1, 4, 2)
plt.title('Sobel Filter')
plt.imshow(sobel_filtered, cmap='gray')

plt.subplot(1, 4, 3)
plt.title('Median Filter')
plt.imshow(median_filtered, cmap='gray')

plt.subplot(1, 4, 4)
plt.title('Sobel + Median Filter')
plt.imshow(sobel_median_filtered, cmap='gray')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread("D:/College Study Stuff/Project/Brain Tumor Web/Datasets/Braintumosparseclassification/Training/glioma_tumor/gg (652).jpg", cv2.IMREAD_GRAYSCALE)

# Apply a Sobel filter
sobel_filtered_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=5)
sobel_filtered_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=5)
sobel_filtered = np.sqrt(sobel_filtered_x**2 + sobel_filtered_y**2)

# Apply a Laplacian filter
laplacian_filtered = cv2.Laplacian(image, cv2.CV_64F)

# Apply both filters in sequence
sobel_laplacian_filtered = np.sqrt(sobel_filtered**2 + laplacian_filtered**2)

# Display the original image and the filtered images
plt.figure(figsize=(12, 4))

plt.subplot(1, 4, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')

plt.subplot(1, 4, 2)
plt.title('Sobel Filter')
plt.imshow(sobel_filtered, cmap='gray')

plt.subplot(1, 4, 3)
plt.title('Laplacian Filter')
plt.imshow(laplacian_filtered, cmap='gray')

plt.subplot(1, 4, 4)
plt.title('Sobel + Laplacian Filter')
plt.imshow(sobel_laplacian_filtered, cmap='gray')

plt.tight_layout()
plt.show()
```
